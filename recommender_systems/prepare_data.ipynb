```python
movies_df = pd.read_csv("/Users/felipepesantez/Documents/development/datasets/recommend_systems/archive/movies.csv")
rating_df = pd.read_csv("/Users/felipepesantez/Documents/development/datasets/recommend_systems/archive/ratings.csv")
```

```python
def set_col_mapping(df, col_type):
    object_columns = df.select_dtypes(include=[col_type]).columns.tolist()
    print(object_columns)
    for col in object_columns:
        res = {}
        for x, value in enumerate(df[col].value_counts().index.tolist()):
            res[value] = x
        print(res)
        df[col] = df[col].map(res)

    return df
```

```python
mv_df = set_col_mapping(movies_df, 'object')
rt_df = set_col_mapping(rating_df, 'object')
```

```python
mv_df.head()
```

```python
merged_df = pd.merge(mv_df, rt_df, on="movieId")
```

```python
merged_df.head()
```

```python
sample_df = merged_df.drop(["title", "genres", "timestamp"], axis=1)
```

```python
sample_df.to_csv("simple_df.csv", index=False)
```
